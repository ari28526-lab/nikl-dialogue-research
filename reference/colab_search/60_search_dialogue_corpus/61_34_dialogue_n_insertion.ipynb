# 61-34. 대화코퍼스 ㄴ/ㄹ삽입 검색

34번 사전검색 결과(ㄴ/ㄹ삽입 후보 단어)가 대화 코퍼스에서 실제로 사용되는 맥락을 검색

## 검색 대상
- **34번 ㄴ/ㄹ삽입**: 합성어/파생어 내부 형태소 경계에서의 ㄴ삽입, ㄹ삽입

## 참고
- 이 노트북은 **단어 내부(word-internal)** ㄴ삽입만 검색
- **구(phrase) 경계** ㄴ삽입 (예: "할 일" → [할릴])은 코퍼스 수준에서 별도 검색 가능
  - 발화의 어절 경계에서 선행 어절 끝 + 후행 어절 시작의 자음/모음 조합으로 탐지

## 입출력
- 입력: 34번 결과 CSV (`n_l_insertion` 포함 파일), 대화 enriched CSV
- 출력: `search_results/dialogue_34_n_insertion_*.csv`
