# 51-34. 서울코퍼스 발음 검색: ㄴ/ㄹ삽입

34번 사전검색 결과(ㄴ/ㄹ삽입 후보)가 서울 코퍼스에서 실제로 어떻게 발음되는지 확인

## 검색 대상
- **단어 내부 ㄴ/ㄹ삽입**: 30번 사전검색 결과 매칭
- **구 경계 ㄴ삽입**: 연속 어절에서 어절① 종성(자음) + 어절② 초성(i/j) → ㄴ삽입 환경

## 입력
- 34번 결과 CSV: `n_l_insertion` 포함 파일
- 서울 enriched: `04_seoul_pword_enriched.csv`

## 출력
- 단어 매칭 CSV: `search_results/seoul_34_n_insertion_*.csv`
- 구 경계 ㄴ삽입 CSV: `search_results/seoul_34_phrase_n_insertion_*.csv`
